```python
import requests
import pandas as pd


url = 'https://api.coingecko.com/api/v3/coins/markets'
params = {
    'vs_currency': 'usd',
    'order': 'market_cap_desc',
    'per_page': 50,
    'page': 1,
    'sparkline': False
}

response = requests.get(url, params=params)
data = response.json()

df = pd.DataFrame(data, columns=[
    'crypto_name', 'symbol', 'current_price', 'market_cap',
    'trading_volume', 'price_change_percentage_24h'
])
print(df)
```

#data analysis

```python
top_5 = df.nlargest(5, 'market_cap')
print("Top 5 Cryptocurrencies by Market Cap:")
print(top_5[['crypto_name', 'market_cap']])

average_price = df['current_price'].mean()
print(f"Average Price of Top 50 Cryptocurrencies: ${average_price:.2f}")

highest_change = df.loc[df['price_change_percentage_24h'].idxmax()]
lowest_change = df.loc[df['price_change_percentage_24h'].idxmin()]

print("Highest 24-Hour Percentage Price Change:")
print(highest_change[['crypto_name', 'price_change_percentage_24h']])

print("Lowest 24-Hour Percentage Price Change:")
print(lowest_change[['crypto_name', 'price_change_percentage_24h']])
```

#live excel sheet

```python
import requests
import pandas as pd
from openpyxl import Workbook
import time

def fetch_data():
    url = 'https://api.coingecko.com/api/v3/coins/markets'
    params = {
        'vs_currency': 'usd',
        'order': 'market_cap_desc',
        'per_page': 50,
        'page': 1,
        'sparkline': False
    }
    response = requests.get(url, params=params)
    data = response.json()
    df = pd.DataFrame(data, columns=[
        'name', 'symbol', 'current_price', 'market_cap',
        'total_volume', 'price_change_percentage_24h'
    ])
    return df

def update_excel(df):
    with pd.ExcelWriter('crypto_data.xlsx', engine='openpyxl') as writer:
        df.to_excel(writer, index=False, sheet_name='Top 50 Cryptocurrencies')

while True:
    df = fetch_data()
    update_excel(df)
    print("Excel sheet updated.")
    time.sleep(300)
```
